```python
print("======================================")
print(" ASISTENTE IA DE AMORTIZACION")
print("======================================")

continuar = "si"

while continuar == "si":

    print("\nHola, soy tu asistente financiero.")
    print("Te ayudare a analizar un prestamo.\n")

    nombre = input("¿Cual es tu nombre?: ")

    print("\nMucho gusto,", nombre)

    prestamo = float(input("\nIngresa el monto del prestamo: "))
    tasa = float(input("Ingresa la tasa de interes anual (%): "))
    años = int(input("¿Cuantos años deseas pagar?: "))

    meses = años * 12
    interes_mensual = tasa / 100 / 12

    print("\nAnalizando informacion...")

    if prestamo <= 0:
        print("El monto del prestamo no es valido.")

    elif tasa <= 0:
        print("La tasa de interes no es valida.")

    elif años <= 0:
        print("La cantidad de años no es valida.")

    else:

        cuota = (prestamo * interes_mensual) / (1 - (1 + interes_mensual) ** (-meses))

        total_pagar = cuota * meses
        intereses = total_pagar - prestamo

        print("\n=========== RESULTADOS ===========")
        print("Monto prestado: ", round(prestamo, 2))
        print("Tasa anual: ", tasa, "%")
        print("Tiempo: ", años, "años")
        print("Pago mensual aproximado: ", round(cuota, 2))
        print("Total pagado: ", round(total_pagar, 2))
        print("Intereses pagados: ", round(intereses, 2))

        print("\n=========== ANALISIS IA ===========")

        if tasa < 5:
            print("La IA detecta una tasa de interes MUY BAJA.")

        elif tasa >= 5 and tasa < 12:
            print("La IA considera que la tasa es NORMAL.")

        else:
            print("La IA detecta una tasa de interes ALTA.")

        if años <= 3:
            print("El tiempo de pago es corto.")
            print("Pagaras menos intereses.")

        elif años > 3 and años <= 10:
            print("El tiempo de pago es moderado.")

        else:
            print("El tiempo de pago es largo.")
            print("Podrias terminar pagando muchos intereses.")

        if intereses > prestamo:
            print("ALERTA: pagaras mas intereses que el dinero prestado.")

        else:
            print("La deuda parece manejable.")

        print("\n=========== RECOMENDACION ===========")

        if cuota > 10000:
            print("La cuota mensual es elevada.")
            print("La IA recomienda buscar un plazo mas largo.")

        elif cuota >= 3000 and cuota <= 10000:
            print("La cuota mensual parece aceptable.")

        else:
            print("La cuota mensual es baja.")
            print("La IA considera que el prestamo es accesible.")

    continuar = input("\n¿Deseas realizar otro analisis? (si/no): ")

print("\nGracias por utilizar el asistente IA.")
print("Programa finalizado.")
```
